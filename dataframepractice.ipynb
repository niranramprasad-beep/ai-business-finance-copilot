```python
# Cell 1 — load the table and show the first 5 rows
import pandas as pd

df = pd.read_csv("superstore.csv", encoding="latin-1")
df.head()
```

```python
# Cell 3 — revenue by region
df.groupby("Region")["Sales"].sum()
```

```python
# Cell 5 — download Apple and Disney prices
import yfinance as yf
stocks = yf.download(["AAPL", "DIS"], period="1y")["Close"]
```

```python
# Cell 6 — Apple vs Disney chart
stocks.plot(title="Apple vs Disney — Last 12 Months")
```

```python
# Cell 7 — daily % change
stocks.pct_change().plot(title="Daily % Change")
```
