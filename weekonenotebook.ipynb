```python
# Cell 1 
import pandas as pd

df = pd.read_csv("superstore.csv", encoding="latin-1")
df.head()
```

```python
# Cell 2 
df["Sales"].sum()
```

```python
# Cell 3    
df.groupby("Region")["Sales"].sum()
```

```python
# Cell 4    
df.groupby("Product Name")["Sales"].sum().nlargest(5)
```

```python
# Cell 5    
import yfinance as yf
stocks = yf.download(["AAPL", "DIS"], period="1y")["Close"]
```

```python
# Cell 6    
stocks.plot(title="Apple vs Disney — Last 12 Months")
```

```python
# Cell 7    
stocks.pct_change().plot(title="Daily % Change")
```

```python
# Cell 8    
df["Order Date"] = pd.to_datetime(df["Order Date"])
df.groupby(df["Order Date"].dt.to_period("M"))["Sales"].sum().plot(title="Monthly Revenue")
```

```python
# Cell 9    
df.groupby("Category")["Profit"].sum().plot(kind="bar", title="Profit by Category")
```

```python
# Cell 10   
growth = (stocks.iloc[-1] / stocks.iloc[0] - 1) * 100
print(growth.round(1))
```
